```python
import os
import json
import yaml
import joblib
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import Adam

warnings.filterwarnings("ignore")

np.random.seed(42)

# ============================================================
# 2. DEFINE PROJECT PATH
# ============================================================

PROJECT_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Budget Allocation Trend Analysis "
    r"and Forecasting Using Machine Learning"
)

DATASET_PATH = os.path.join(
    PROJECT_PATH,
    "RS_Session_252_AU_600.csv"
)

os.makedirs(PROJECT_PATH, exist_ok=True)

print("Project Directory:", PROJECT_PATH)

# ============================================================
# 3. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

df.columns = df.columns.str.strip()

# Extract financial years and corresponding allocation values
# from the dataset.
#
# This parser supports:
# 1. Long format: Financial Year | Budget Allocation
# 2. Wide format: 2017-18 | 2018-19 | 2019-20
#
# It does not assume that the first two columns contain
# the year and allocation.

import re

year_pattern = re.compile(
    r"(?:19|20)\d{2}\s*[-–/]\s*(?:\d{2}|(?:19|20)\d{2})"
)

def parse_numeric(series):
    return pd.to_numeric(
        series.astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("₹", "", regex=False)
        .str.replace(r"(?i)\s*crores?\s*", "", regex=True)
        .str.strip(),
        errors="coerce"
    )

def extract_year(value):
    match = year_pattern.search(str(value))

    if match:
        return int(match.group(0)[:4])

    return np.nan

# Detect wide-format year columns
year_columns = [
    col for col in df.columns
    if year_pattern.search(str(col))
]

if year_columns:

    # If multiple rows exist, use the first row containing
    # the maximum number of valid allocation values.
    numeric_year_data = df[year_columns].apply(parse_numeric)

    best_row = numeric_year_data.notna().sum(axis=1).idxmax()

    records = []

    for col in year_columns:
        value = numeric_year_data.loc[best_row, col]

        if pd.notna(value):
            records.append({
                "Year": extract_year(col),
                "Budget_Allocation": float(value)
            })

    data = pd.DataFrame(records)

else:

    # Detect long-format columns
    year_candidates = [
        col for col in df.columns
        if "year" in col.lower()
    ]

    allocation_candidates = [
        col for col in df.columns
        if any(
            word in col.lower()
            for word in [
                "allocation",
                "budget",
                "amount",
                "fund"
            ]
        )
    ]

    if not year_candidates or not allocation_candidates:
        raise ValueError(
            "Unable to identify financial year and budget "
            "allocation columns. Please inspect df.columns."
        )

    year_col = year_candidates[0]

    allocation_col = next(
        (
            col for col in allocation_candidates
            if col != year_col
        ),
        None
    )

    if allocation_col is None:
        raise ValueError(
            "No separate budget allocation column found."
        )

    data = pd.DataFrame({
        "Year": df[year_col].apply(extract_year),
        "Budget_Allocation": parse_numeric(
            df[allocation_col]
        )
    })

data = data.dropna()

# Ensure one allocation value per year
if data["Year"].duplicated().any():
    raise ValueError(
        "Multiple allocations found for the same year. "
        "Select a specific budget category before training."
    )

data = data.sort_values("Year").reset_index(drop=True)

if len(data) < 2:
    raise ValueError(
        "At least two valid historical records are required."
    )

print("\nCleaned Dataset:")
print(data)

# Save cleaned dataset
data.to_csv(
    os.path.join(PROJECT_PATH, "cleaned_dataset.csv"),
    index=False
)

# ============================================================
# 5. FEATURE SELECTION
# ============================================================

X = data[["Year"]].values.astype(float)

y = data["Budget_Allocation"].values.astype(float)

print("\nInput Features:")
print(X)

print("\nTarget Values:")
print(y)

# ============================================================
# 6. FEATURE SCALING
# ============================================================

x_scaler = StandardScaler()
y_scaler = StandardScaler()

X_scaled = x_scaler.fit_transform(X)

y_scaled = y_scaler.fit_transform(
    y.reshape(-1, 1)
)

# Save scalers
joblib.dump(
    x_scaler,
    os.path.join(PROJECT_PATH, "x_scaler.pkl")
)

joblib.dump(
    y_scaler,
    os.path.join(PROJECT_PATH, "y_scaler.pkl")
)

print("\nFeature Scaling Completed.")

# ============================================================
# 7. LINEAR REGRESSION MODEL
# ============================================================

lr_model = LinearRegression()

lr_model.fit(X_scaled, y)

lr_predictions = lr_model.predict(X_scaled)

# Save Linear Regression model
joblib.dump(
    lr_model,
    os.path.join(
        PROJECT_PATH,
        "budget_prediction_model.pkl"
    )
)

print("\nLinear Regression Model Trained.")

# ============================================================
# 8. ARTIFICIAL NEURAL NETWORK MODEL
# ============================================================

ann_model = Sequential([
    Input(shape=(1,)),

    Dense(
        32,
        activation="relu"
    ),

    Dense(
        16,
        activation="relu"
    ),

    Dense(
        8,
        activation="relu"
    ),

    Dense(
        1,
        activation="linear"
    )
])

ann_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

print("\nANN Model Architecture:")
ann_model.summary()

history = ann_model.fit(
    X_scaled,
    y_scaled,
    epochs=300,
    batch_size=1,
    verbose=0
)

print("\nANN Model Training Completed.")

# ============================================================
# 9. SAVE ANN MODEL AS H5
# ============================================================

h5_path = os.path.join(
    PROJECT_PATH,
    "budget_prediction_model.h5"
)

ann_model.save(h5_path)

print("\nANN Model Saved:", h5_path)

# ============================================================
# 10. GENERATE ANN PREDICTIONS
# ============================================================

ann_predictions_scaled = ann_model.predict(
    X_scaled,
    verbose=0
)

ann_predictions = y_scaler.inverse_transform(
    ann_predictions_scaled
).flatten()

# ============================================================
# 11. MODEL EVALUATION
# ============================================================

def evaluate_model(y_true, y_pred):

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    mse = mean_squared_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_true,
        y_pred
    )

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2_Score": float(r2)
    }


lr_metrics = evaluate_model(
    y,
    lr_predictions
)

ann_metrics = evaluate_model(
    y,
    ann_predictions
)

print("\nLinear Regression Training Metrics:")
print(lr_metrics)

print("\nANN Training Metrics:")
print(ann_metrics)

print(
    "\nWARNING: These are in-sample training metrics, "
    "not independent test accuracy."
)

# ============================================================
# 12. FUTURE BUDGET PREDICTION
# ============================================================

last_year = int(data["Year"].max())

future_years = np.arange(
    last_year + 1,
    last_year + 7
).reshape(-1, 1)

future_scaled = x_scaler.transform(
    future_years
)

# Linear Regression Forecast
future_lr_predictions = lr_model.predict(
    future_scaled
)

# ANN Forecast
future_ann_scaled = ann_model.predict(
    future_scaled,
    verbose=0
)

future_ann_predictions = y_scaler.inverse_transform(
    future_ann_scaled
).flatten()

future_results = pd.DataFrame({
    "Year": future_years.flatten(),
    "Linear_Regression_Prediction":
        future_lr_predictions,
    "ANN_Prediction":
        future_ann_predictions
})

print("\nFuture Budget Predictions:")
print(future_results)

# Save predictions
future_results.to_csv(
    os.path.join(
        PROJECT_PATH,
        "future_budget_predictions.csv"
    ),
    index=False
)

# ============================================================
# 13. SAVE MODEL CONFIGURATION AS YAML
# ============================================================

model_config = {

    "project_name":
        "AI-Based Government Budget Allocation "
        "Trend Analysis and Forecasting",

    "dataset":
        "RS_Session_252_AU_600.csv",

    "target_column":
        "Budget_Allocation",

    "feature_columns": [
        "Year"
    ],

    "models": {

        "linear_regression": {
            "algorithm": "LinearRegression",
            "scaling": "StandardScaler"
        },

        "ann": {
            "algorithm": "Artificial Neural Network",
            "hidden_layers": [32, 16, 8],
            "activation": "relu",
            "optimizer": "Adam",
            "learning_rate": 0.001,
            "epochs": 300,
            "batch_size": 1
        }
    },

    "forecast": {
        "future_years": 6
    },

    "evaluation": {
        "metrics": [
            "MAE",
            "MSE",
            "RMSE",
            "R2_Score"
        ],
        "type": "in_sample_training"
    }
}

yaml_path = os.path.join(
    PROJECT_PATH,
    "model_config.yaml"
)

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        model_config,
        file,
        sort_keys=False
    )

print("\nYAML Configuration Saved.")

# ============================================================
# 14. SAVE MODEL METADATA AS JSON
# ============================================================

model_metadata = {

    "project_name":
        "AI-Based Government Budget Allocation "
        "Trend Analysis and Forecasting",

    "dataset_name":
        "RS_Session_252_AU_600.csv",

    "total_records":
        int(len(data)),

    "features": [
        "Year"
    ],

    "target":
        "Budget_Allocation",

    "models": [
        "Linear Regression",
        "Artificial Neural Network"
    ],

    "linear_regression_metrics":
        lr_metrics,

    "ann_metrics":
        ann_metrics,

    "model_files": {
        "ann_model":
            "budget_prediction_model.h5",

        "linear_regression_model":
            "budget_prediction_model.pkl",

        "x_scaler":
            "x_scaler.pkl",

        "y_scaler":
            "y_scaler.pkl"
    },

    "evaluation_type":
        "Training data only",

    "limitations":
        "Small historical dataset; predictions "
        "are illustrative and not validated forecasts."
}

json_path = os.path.join(
    PROJECT_PATH,
    "model_metadata.json"
)

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        model_metadata,
        file,
        indent=4
    )

print("\nJSON Metadata Saved.")

# ============================================================
# 15. SAVE FUTURE PREDICTIONS AS JSON
# ============================================================

future_json_path = os.path.join(
    PROJECT_PATH,
    "future_budget_predictions.json"
)

with open(
    future_json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        future_results.to_dict(
            orient="records"
        ),
        file,
        indent=4
    )

print("\nFuture Predictions JSON Saved.")

# ============================================================
# 16. BUDGET ALLOCATION TREND GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    data["Year"],
    data["Budget_Allocation"],
    marker="o",
    linewidth=2,
    label="Historical Budget"
)

plt.title(
    "Government Budget Allocation Trend"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Budget Allocation")
plt.legend()
plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_PATH,
        "budget_trend.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 17. FUTURE BUDGET FORECAST GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    data["Year"],
    data["Budget_Allocation"],
    marker="o",
    label="Historical Budget"
)

plt.plot(
    future_results["Year"],
    future_results[
        "Linear_Regression_Prediction"
    ],
    marker="s",
    linestyle="--",
    label="Linear Regression Forecast"
)

plt.plot(
    future_results["Year"],
    future_results["ANN_Prediction"],
    marker="^",
    linestyle="--",
    label="ANN Forecast"
)

plt.title(
    "Government Budget Allocation Forecast"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Predicted Budget Allocation")

plt.legend()
plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_PATH,
        "budget_forecast.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 18. ANN TRAINING LOSS GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["loss"],
    label="ANN Training Loss"
)

plt.title("ANN Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")
plt.legend()
plt.grid(True)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_PATH,
        "ann_training_loss.png"
    ),
    dpi=300
)

plt.show()

# ============================================================
# 19. DISPLAY ALL GENERATED FILES
# ============================================================

print("\n" + "=" * 60)
print("PROJECT EXECUTION COMPLETED")
print("=" * 60)

print("\nGenerated Files:")

for filename in sorted(os.listdir(PROJECT_PATH)):

    print(filename)

print("\nAll results saved in:")
print(PROJECT_PATH)

print("\n" + "=" * 60)
```
