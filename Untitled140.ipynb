```python
import os
import re
import json
import yaml
import joblib
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

np.random.seed(42)
sns.set_theme(style="whitegrid")


# ============================================================
# 2. PROJECT PATH
# ============================================================

PROJECT_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Budget Allocation Trend Analysis "
    r"and Forecasting Using Machine Learning"
)

DATASET_PATH = os.path.join(
    PROJECT_PATH,
    "RS_Session_252_AU_600.csv"
)

os.makedirs(PROJECT_PATH, exist_ok=True)

PREFIX = "ais_"


def save_path(filename):
    return os.path.join(
        PROJECT_PATH,
        PREFIX + filename
    )


print("=" * 65)
print("ARTIFICIAL IMMUNE SYSTEM - BUDGET FORECASTING")
print("=" * 65)

print("\nProject Directory:")
print(PROJECT_PATH)


# ============================================================
# 3. LOAD DATASET
# ============================================================

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.astype(str).str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())


# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

year_pattern = re.compile(
    r"((?:19|20)\d{2})\s*[-–/]\s*"
    r"(?:\d{2}|(?:19|20)\d{2})"
)


def extract_year(value):

    text = str(value).strip()

    match = year_pattern.search(text)

    if match:
        return int(match.group(1))

    if re.fullmatch(r"(?:19|20)\d{2}", text):
        return int(text)

    return np.nan


def convert_numeric(series):

    cleaned = (
        series.astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("₹", "", regex=False)
        .str.replace(
            r"(?i)\s*crores?\s*",
            "",
            regex=True
        )
        .str.strip()
    )

    return pd.to_numeric(
        cleaned,
        errors="coerce"
    )


year_columns = [
    col for col in df.columns
    if year_pattern.search(str(col))
]

if len(year_columns) >= 2:

    numeric_data = df[year_columns].apply(
        convert_numeric
    )

    # Select the row containing the most valid values.
    # For multiple budget categories, explicitly filter
    # the required category instead.

    best_row = numeric_data.notna().sum(
        axis=1
    ).idxmax()

    records = []

    for col in year_columns:

        value = numeric_data.loc[
            best_row,
            col
        ]

        if pd.notna(value):

            records.append({
                "Year": extract_year(col),
                "Budget_Allocation": float(value)
            })

    data = pd.DataFrame(records)

else:

    year_candidates = [
        col for col in df.columns
        if "year" in col.lower()
    ]

    budget_candidates = [
        col for col in df.columns
        if any(
            word in col.lower()
            for word in [
                "allocation",
                "budget",
                "amount",
                "fund"
            ]
        )
    ]

    if not year_candidates or not budget_candidates:
        raise ValueError(
            "Unable to detect year and budget columns. "
            "Inspect the CSV column names."
        )

    year_col = year_candidates[0]

    budget_col = next(
        (
            col for col in budget_candidates
            if col != year_col
        ),
        None
    )

    if budget_col is None:
        raise ValueError(
            "A separate budget column is required."
        )

    data = pd.DataFrame({
        "Year": df[year_col].apply(
            extract_year
        ),
        "Budget_Allocation": convert_numeric(
            df[budget_col]
        )
    })


data = data.dropna()

data = data.sort_values(
    "Year"
).reset_index(drop=True)

if data["Year"].duplicated().any():
    raise ValueError(
        "Multiple budget values exist for the same year. "
        "Filter to one budget category before training."
    )

if len(data) < 2:
    raise ValueError(
        "At least two historical records are required."
    )

data.to_csv(
    save_path("cleaned_dataset.csv"),
    index=False
)

print("\nCleaned Dataset:")
print(data)


# ============================================================
# 5. FEATURE PREPARATION
# ============================================================

years = data["Year"].to_numpy(
    dtype=float
)

y = data["Budget_Allocation"].to_numpy(
    dtype=float
)

# Center and scale years for numerical stability.

year_center = float(np.mean(years))

year_scale = float(np.std(years))

if year_scale == 0:
    year_scale = 1.0

X_scaled = (
    years - year_center
) / year_scale

# Scale target values.

target_center = float(np.mean(y))

target_scale = float(np.std(y))

if target_scale == 0:
    target_scale = 1.0

y_scaled = (
    y - target_center
) / target_scale


# ============================================================
# 6. AIS MODEL REPRESENTATION
# ============================================================

# The AIS algorithm optimizes a linear trend:
#
# y_scaled = a + b*x_scaled
#
# Each antibody represents two coefficients:
#
# antibody = [a, b]
#
# A linear model is used because the dataset contains
# very few historical observations.


def predict_scaled(
    antibodies,
    x_values
):

    antibodies = np.asarray(
        antibodies,
        dtype=float
    )

    x_values = np.asarray(
        x_values,
        dtype=float
    )

    return (
        antibodies[..., 0, None]
        +
        antibodies[..., 1, None]
        * x_values
    )


def fitness_function(
    population,
    x_values,
    targets
):

    predictions = predict_scaled(
        population,
        x_values
    )

    errors = (
        predictions
        - targets
    ) ** 2

    return np.mean(
        errors,
        axis=-1
    )


# ============================================================
# 7. AIS CLONAL SELECTION ALGORITHM
# ============================================================

def artificial_immune_system(
    x_values,
    targets,
    population_size=50,
    generations=200,
    clone_factor=5,
    mutation_rate=0.20,
    elite_fraction=0.20,
    random_state=42
):

    rng = np.random.default_rng(
        random_state
    )

    # Each antibody contains two model coefficients.
    population = rng.uniform(
        low=-3,
        high=3,
        size=(population_size, 2)
    )

    convergence_history = []

    global_best = None
    global_best_fitness = np.inf

    for generation in range(
        generations
    ):

        fitness = fitness_function(
            population,
            x_values,
            targets
        )

        sorted_indices = np.argsort(
            fitness
        )

        population = population[
            sorted_indices
        ]

        fitness = fitness[
            sorted_indices
        ]

        # Update global best antibody.
        if fitness[0] < global_best_fitness:

            global_best_fitness = float(
                fitness[0]
            )

            global_best = population[
                0
            ].copy()

        convergence_history.append(
            global_best_fitness
        )

        # Select the best antibodies.
        elite_count = max(
            2,
            int(
                population_size
                * elite_fraction
            )
        )

        selected = population[
            :elite_count
        ]

        selected_fitness = fitness[
            :elite_count
        ]

        # Generate clones.
        clones = np.repeat(
            selected,
            clone_factor,
            axis=0
        )

        clone_fitness = np.repeat(
            selected_fitness,
            clone_factor
        )

        # Better antibodies receive smaller mutations.
        normalized_fitness = (
            clone_fitness
            /
            (
                np.max(clone_fitness)
                + 1e-12
            )
        )

        mutation_strength = (
            mutation_rate
            *
            (
                0.1
                + normalized_fitness
            )
        )

        mutations = rng.normal(
            loc=0,
            scale=mutation_strength[:, None],
            size=clones.shape
        )

        mutated_clones = (
            clones + mutations
        )

        # Introduce random antibodies for diversity.
        random_count = max(
            2,
            population_size // 10
        )

        random_antibodies = rng.uniform(
            -3,
            3,
            size=(random_count, 2)
        )

        # Combine candidates.
        candidates = np.vstack([
            population,
            mutated_clones,
            random_antibodies,
            global_best.reshape(1, -1)
        ])

        candidate_fitness = fitness_function(
            candidates,
            x_values,
            targets
        )

        best_indices = np.argsort(
            candidate_fitness
        )[:population_size]

        population = candidates[
            best_indices
        ]

        if (
            generation + 1
        ) % 20 == 0:

            print(
                f"Generation {generation + 1:3d} "
                f"| Best MSE: "
                f"{global_best_fitness:.8f}"
            )

    # Check final population.
    final_fitness = fitness_function(
        population,
        x_values,
        targets
    )

    final_index = np.argmin(
        final_fitness
    )

    if final_fitness[final_index] < global_best_fitness:

        global_best = population[
            final_index
        ].copy()

        global_best_fitness = float(
            final_fitness[final_index]
        )

    return (
        global_best,
        global_best_fitness,
        convergence_history
    )


# ============================================================
# 8. TRAIN AIS MODEL
# ============================================================

AIS_POPULATION = 50
AIS_GENERATIONS = 200
AIS_CLONE_FACTOR = 5
AIS_MUTATION_RATE = 0.20

print("\nTraining AIS Model...\n")

(
    best_antibody,
    best_fitness,
    convergence_history
) = artificial_immune_system(

    X_scaled,
    y_scaled,

    population_size=AIS_POPULATION,

    generations=AIS_GENERATIONS,

    clone_factor=AIS_CLONE_FACTOR,

    mutation_rate=AIS_MUTATION_RATE
)

print("\nAIS Training Completed.")

print("\nBest Antibody:")
print(best_antibody)

print("\nBest Fitness:")
print(best_fitness)


# ============================================================
# 9. AIS PREDICTIONS
# ============================================================

ais_scaled_predictions = predict_scaled(
    best_antibody,
    X_scaled
)

ais_predictions = (
    ais_scaled_predictions
    * target_scale
    + target_center
)


# ============================================================
# 10. BASELINE LINEAR REGRESSION
# ============================================================

baseline_model = LinearRegression()

baseline_model.fit(
    X_scaled.reshape(-1, 1),
    y
)

baseline_predictions = baseline_model.predict(
    X_scaled.reshape(-1, 1)
)


# ============================================================
# 11. MODEL EVALUATION
# ============================================================

def evaluate_model(
    actual,
    predicted
):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual,
        predicted
    )

    nonzero = actual != 0

    if np.any(nonzero):

        mape = np.mean(
            np.abs(
                (
                    actual[nonzero]
                    - predicted[nonzero]
                )
                / actual[nonzero]
            )
        ) * 100

        accuracy = 100 - mape

    else:

        mape = np.nan
        accuracy = np.nan

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": float(mape),
        "Accuracy_Percentage": float(accuracy)
    }


ais_metrics = evaluate_model(
    y,
    ais_predictions
)

baseline_metrics = evaluate_model(
    y,
    baseline_predictions
)

print("\nAIS Training Metrics:")
print(ais_metrics)

print("\nBaseline Training Metrics:")
print(baseline_metrics)

print(
    "\nNOTE: All metrics are calculated on historical "
    "training data, not an independent test set."
)


# ============================================================
# 12. SAVE AIS MODEL
# ============================================================

ais_model_data = {

    "algorithm": "AIS_CLONALG",

    "model_type": "Linear Regression",

    "best_antibody": best_antibody,

    "year_center": year_center,

    "year_scale": year_scale,

    "target_center": target_center,

    "target_scale": target_scale,

    "best_fitness": best_fitness
}

joblib.dump(
    ais_model_data,
    save_path("model.pkl")
)

print("\nAIS Model Saved.")


# ============================================================
# 13. SAVE AIS CONFIGURATION YAML
# ============================================================

ais_config = {

    "project":
        "Government Budget Allocation Forecasting",

    "algorithm":
        "Artificial Immune System",

    "optimization":
        "Clonal Selection Algorithm",

    "population_size":
        AIS_POPULATION,

    "generations":
        AIS_GENERATIONS,

    "clone_factor":
        AIS_CLONE_FACTOR,

    "mutation_rate":
        AIS_MUTATION_RATE,

    "model":
        "Linear Trend",

    "features": [
        "Year"
    ],

    "target":
        "Budget_Allocation",

    "evaluation":
        "Training Data Only"
}

with open(
    save_path("model_config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        ais_config,
        file,
        sort_keys=False
    )

print("\nAIS YAML Saved.")


# ============================================================
# 14. SAVE AIS METADATA JSON
# ============================================================

ais_metadata = {

    "algorithm":
        "Artificial Immune System",

    "optimization":
        "CLONALG",

    "dataset":
        "RS_Session_252_AU_600.csv",

    "total_records":
        int(len(data)),

    "best_antibody":
        best_antibody.tolist(),

    "best_fitness":
        float(best_fitness),

    "metrics":
        ais_metrics,

    "limitations":
        "Small dataset; forecasts are experimental."
}

with open(
    save_path("model_metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        ais_metadata,
        file,
        indent=4
    )

print("\nAIS Metadata Saved.")


# ============================================================
# 15. AIS ACCURACY GRAPH
# ============================================================

# The displayed metric is 100 - training MAPE.
# It is not classification accuracy or test accuracy.

models = [
    "Baseline Linear Regression",
    "AIS Optimized Regression"
]

accuracies = [
    baseline_metrics["Accuracy_Percentage"],
    ais_metrics["Accuracy_Percentage"]
]

plt.figure(figsize=(10, 6))

bars = plt.bar(
    models,
    accuracies,
    color=["steelblue", "darkorange"],
    width=0.5
)

for bar, value in zip(
    bars,
    accuracies
):

    if np.isfinite(value):

        plt.text(
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height(),
            f"{value:.2f}%",
            ha="center",
            va="bottom",
            fontsize=12
        )

plt.title(
    "AIS Training Fit Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("100 - Training MAPE (%)")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    save_path("accuracy_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAIS Accuracy Graph Saved.")


# ============================================================
# 16. AIS CORRELATION HEATMAP
# ============================================================

heatmap_df = data.copy()

heatmap_df["AIS_Prediction"] = (
    ais_predictions
)

heatmap_df["Budget_Growth"] = (
    heatmap_df[
        "Budget_Allocation"
    ].pct_change() * 100
)

correlation_matrix = heatmap_df[
    [
        "Year",
        "Budget_Allocation",
        "AIS_Prediction",
        "Budget_Growth"
    ]
].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title(
    "AIS Budget Allocation Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    save_path("heatmap.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAIS Heatmap Saved.")


# ============================================================
# 17. AIS MODEL COMPARISON GRAPH
# ============================================================

comparison_df = pd.DataFrame({

    "Model": models,

    "MAE": [
        baseline_metrics["MAE"],
        ais_metrics["MAE"]
    ],

    "MSE": [
        baseline_metrics["MSE"],
        ais_metrics["MSE"]
    ],

    "RMSE": [
        baseline_metrics["RMSE"],
        ais_metrics["RMSE"]
    ],

    "R2": [
        baseline_metrics["R2"],
        ais_metrics["R2"]
    ],

    "MAPE": [
        baseline_metrics["MAPE"],
        ais_metrics["MAPE"]
    ]
})

comparison_df.to_csv(
    save_path("model_comparison.csv"),
    index=False
)

x_positions = np.arange(
    len(models)
)

width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x_positions - width / 2,
    comparison_df["MAE"],
    width,
    label="MAE",
    color="royalblue"
)

plt.bar(
    x_positions + width / 2,
    comparison_df["RMSE"],
    width,
    label="RMSE",
    color="darkorange"
)

plt.xticks(
    x_positions,
    models
)

plt.title(
    "AIS vs Linear Regression - Training Error",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("Error")

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    save_path("comparison_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAIS Comparison Graph Saved.")


# ============================================================
# 18. AIS RESULT CSV
# ============================================================

result_df = pd.DataFrame({

    "Year": years.astype(int),

    "Actual_Budget": y,

    "AIS_Prediction":
        ais_predictions,

    "Baseline_Prediction":
        baseline_predictions
})

result_df["AIS_Absolute_Error"] = np.abs(
    result_df["Actual_Budget"]
    - result_df["AIS_Prediction"]
)

result_df["Baseline_Absolute_Error"] = np.abs(
    result_df["Actual_Budget"]
    - result_df["Baseline_Prediction"]
)

result_df.to_csv(
    save_path("result.csv"),
    index=False
)

print("\nAIS Result CSV:")
print(result_df)


# ============================================================
# 19. AIS RESULT GRAPH
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(
    result_df["Year"],
    result_df["Actual_Budget"],
    marker="o",
    linewidth=3,
    color="black",
    label="Actual Budget"
)

plt.plot(
    result_df["Year"],
    result_df["AIS_Prediction"],
    marker="s",
    linestyle="--",
    linewidth=2,
    color="darkorange",
    label="AIS Prediction"
)

plt.plot(
    result_df["Year"],
    result_df["Baseline_Prediction"],
    marker="^",
    linestyle=":",
    linewidth=2,
    color="royalblue",
    label="Baseline Prediction"
)

plt.title(
    "AIS Actual vs Predicted Budget Allocation",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Budget Allocation")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    save_path("result_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAIS Result Graph Saved.")


# ============================================================
# 20. AIS FUTURE PREDICTION
# ============================================================

last_year = int(
    np.max(years)
)

future_years = np.arange(
    last_year + 1,
    last_year + 7
)

future_scaled = (
    future_years - year_center
) / year_scale

future_ais_scaled = predict_scaled(
    best_antibody,
    future_scaled
)

future_ais = (
    future_ais_scaled
    * target_scale
    + target_center
)

future_baseline = baseline_model.predict(
    future_scaled.reshape(-1, 1)
)

prediction_df = pd.DataFrame({

    "Year": future_years,

    "AIS_Prediction":
        future_ais,

    "Baseline_Prediction":
        future_baseline
})

prediction_df["Financial_Year"] = (
    prediction_df["Year"].astype(str)
    + "-"
    + (
        (
            prediction_df["Year"] + 1
        ) % 100
    ).astype(str).str.zfill(2)
)

prediction_df = prediction_df[
    [
        "Financial_Year",
        "Year",
        "AIS_Prediction",
        "Baseline_Prediction"
    ]
]

prediction_df.to_csv(
    save_path("prediction.csv"),
    index=False
)

print("\nAIS Future Predictions:")
print(prediction_df)


# ============================================================
# 21. AIS PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(
    years,
    y,
    marker="o",
    linewidth=3,
    color="black",
    label="Historical Budget"
)

plt.plot(
    future_years,
    future_ais,
    marker="s",
    linestyle="--",
    linewidth=2,
    color="darkorange",
    label="AIS Forecast"
)

plt.plot(
    future_years,
    future_baseline,
    marker="^",
    linestyle=":",
    linewidth=2,
    color="royalblue",
    label="Linear Regression Forecast"
)

plt.axvline(
    x=last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)

plt.title(
    "AIS Government Budget Allocation Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Predicted Budget Allocation")

plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    save_path("prediction_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAIS Prediction Graph Saved.")


# ============================================================
# 22. AIS CONVERGENCE CSV
# ============================================================

convergence_df = pd.DataFrame({

    "Generation": np.arange(
        1,
        len(convergence_history) + 1
    ),

    "Best_Fitness":
        convergence_history
})

convergence_df.to_csv(
    save_path("convergence.csv"),
    index=False
)

print("\nAIS Convergence Data:")
print(convergence_df.tail())


# ============================================================
# 23. AIS CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    convergence_df["Generation"],
    convergence_df["Best_Fitness"],
    color="purple",
    linewidth=2
)

plt.title(
    "AIS Clonal Selection Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Generation")
plt.ylabel("Best Fitness (MSE)")

plt.grid(True, alpha=0.3)

plt.tight_layout()

plt.savefig(
    save_path("convergence_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAIS Convergence Graph Saved.")


# ============================================================
# 24. SAVE EVALUATION METRICS
# ============================================================

evaluation_data = {

    "algorithm":
        "Artificial Immune System",

    "optimization":
        "Clonal Selection",

    "dataset_records":
        int(len(data)),

    "AIS_Metrics":
        ais_metrics,

    "Baseline_Metrics":
        baseline_metrics,

    "Best_Fitness":
        float(best_fitness),

    "Evaluation_Type":
        "In-sample training",

    "Warning":
        "Forecasts are illustrative because "
        "the historical dataset is very small."
}

with open(
    save_path("evaluation_metrics.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        evaluation_data,
        file,
        indent=4
    )

print("\nAIS Evaluation Metrics Saved.")


# ============================================================
# 25. FINAL OUTPUT SUMMARY
# ============================================================

generated_files = [

    "ais_model.pkl",
    "ais_model_config.yaml",
    "ais_model_metadata.json",

    "ais_cleaned_dataset.csv",

    "ais_accuracy_graph.png",
    "ais_heatmap.png",
    "ais_comparison_graph.png",

    "ais_model_comparison.csv",

    "ais_result.csv",
    "ais_result_graph.png",

    "ais_prediction.csv",
    "ais_prediction_graph.png",

    "ais_convergence.csv",
    "ais_convergence_graph.png",

    "ais_evaluation_metrics.json"
]

print("\n" + "=" * 65)
print("AIS PROJECT EXECUTION COMPLETED")
print("=" * 65)

for filename in generated_files:

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    if os.path.exists(filepath):

        print("SUCCESS:", filename)

    else:

        print("MISSING:", filename)

print("\nAll AIS results saved in:")
print(PROJECT_PATH)

print("\n" + "=" * 65)
```
