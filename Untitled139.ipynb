```python
import os
import re
import json
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import (
    Sequential,
    load_model
)

from tensorflow.keras.layers import (
    Input,
    Dense
)

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

np.random.seed(42)

try:
    import tensorflow as tf
    tf.random.set_seed(42)
except ImportError:
    raise ImportError(
        "TensorFlow is required. "
        "Install it using pip install tensorflow"
    )

sns.set_theme(style="whitegrid")


# ============================================================
# 2. PROJECT DIRECTORY
# ============================================================

PROJECT_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Budget Allocation Trend Analysis "
    r"and Forecasting Using Machine Learning"
)

DATASET_PATH = os.path.join(
    PROJECT_PATH,
    "RS_Session_252_AU_600.csv"
)

os.makedirs(
    PROJECT_PATH,
    exist_ok=True
)

print("=" * 65)
print("GOVERNMENT BUDGET ALLOCATION FORECASTING")
print("=" * 65)

print("\nProject Directory:")
print(PROJECT_PATH)


# ============================================================
# 3. LOAD DATASET
# ============================================================

if not os.path.exists(DATASET_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.astype(str).str.strip()

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())


# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

# Supports:
# 1. Long format:
#    Financial Year | Budget Allocation
#
# 2. Wide format:
#    2017-18 | 2018-19 | 2019-20
#
# For datasets containing multiple categories, select
# the correct category before model training.

year_pattern = re.compile(
    r"((?:19|20)\d{2})\s*[-–/]\s*"
    r"(?:\d{2}|(?:19|20)\d{2})"
)


def extract_year(value):

    text = str(value).strip()

    match = year_pattern.search(text)

    if match:
        return int(match.group(1))

    # Also support a single four-digit year
    single_year = re.fullmatch(
        r"(?:19|20)\d{2}",
        text
    )

    if single_year:
        return int(text)

    return np.nan


def convert_numeric(series):

    cleaned = (
        series.astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("₹", "", regex=False)
        .str.replace(
            r"(?i)\s*crores?\s*",
            "",
            regex=True
        )
        .str.strip()
    )

    return pd.to_numeric(
        cleaned,
        errors="coerce"
    )


# Detect financial-year columns
year_columns = [
    col for col in df.columns
    if year_pattern.search(str(col))
]

if len(year_columns) >= 2:

    numeric_data = df[year_columns].apply(
        convert_numeric
    )

    # Choose the row with the most valid yearly values.
    # If multiple budget categories exist, replace this
    # selection with an explicit category filter.

    best_row = numeric_data.notna().sum(
        axis=1
    ).idxmax()

    records = []

    for col in year_columns:

        value = numeric_data.loc[
            best_row,
            col
        ]

        if pd.notna(value):

            records.append({
                "Year": extract_year(col),
                "Budget_Allocation": float(value)
            })

    data = pd.DataFrame(records)

else:

    year_candidates = [
        col for col in df.columns
        if "year" in col.lower()
    ]

    budget_candidates = [
        col for col in df.columns
        if any(
            word in col.lower()
            for word in [
                "allocation",
                "budget",
                "amount",
                "fund"
            ]
        )
    ]

    if not year_candidates:
        raise ValueError(
            "Financial year column not found."
        )

    if not budget_candidates:
        raise ValueError(
            "Budget allocation column not found."
        )

    year_col = year_candidates[0]

    budget_col = next(
        (
            col for col in budget_candidates
            if col != year_col
        ),
        None
    )

    if budget_col is None:
        raise ValueError(
            "Separate budget column not found."
        )

    data = pd.DataFrame({

        "Year": df[year_col].apply(
            extract_year
        ),

        "Budget_Allocation": convert_numeric(
            df[budget_col]
        )
    })


data = data.dropna()

data = data.sort_values(
    "Year"
).reset_index(drop=True)

if data["Year"].duplicated().any():

    raise ValueError(
        "Multiple budget records found for the same year. "
        "Please filter the dataset to one budget category."
    )

if len(data) < 2:

    raise ValueError(
        "At least two historical records are required."
    )

print("\nCleaned Dataset:")
print(data)

data.to_csv(
    os.path.join(
        PROJECT_PATH,
        "cleaned_dataset.csv"
    ),
    index=False
)


# ============================================================
# 5. FEATURE SELECTION
# ============================================================

X = data[["Year"]].values.astype(float)

y = data[
    "Budget_Allocation"
].values.astype(float)

print("\nInput Features:")
print(X)

print("\nTarget Values:")
print(y)


# ============================================================
# 6. FEATURE SCALING
# ============================================================

x_scaler_path = os.path.join(
    PROJECT_PATH,
    "x_scaler.pkl"
)

y_scaler_path = os.path.join(
    PROJECT_PATH,
    "y_scaler.pkl"
)

lr_model_path = os.path.join(
    PROJECT_PATH,
    "budget_prediction_model.pkl"
)

ann_model_path = os.path.join(
    PROJECT_PATH,
    "budget_prediction_model.h5"
)


# Fit fresh scalers if saved models are incomplete.
saved_files_exist = all(
    os.path.exists(path)
    for path in [
        x_scaler_path,
        y_scaler_path,
        lr_model_path,
        ann_model_path
    ]
)

if saved_files_exist:

    print("\nLoading existing models...")

    try:

        x_scaler = joblib.load(
            x_scaler_path
        )

        y_scaler = joblib.load(
            y_scaler_path
        )

        lr_model = joblib.load(
            lr_model_path
        )

        ann_model = load_model(
            ann_model_path,
            compile=False
        )

        models_loaded = True

    except Exception as error:

        print(
            "Saved models could not be loaded:",
            error
        )

        models_loaded = False

else:

    models_loaded = False


if not models_loaded:

    print("\nTraining new models...")

    x_scaler = StandardScaler()
    y_scaler = StandardScaler()

    X_scaled = x_scaler.fit_transform(X)

    y_scaled = y_scaler.fit_transform(
        y.reshape(-1, 1)
    )

    # Linear Regression
    lr_model = LinearRegression()

    lr_model.fit(
        X_scaled,
        y
    )

    # ANN
    ann_model = Sequential([

        Input(shape=(1,)),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            16,
            activation="relu"
        ),

        Dense(
            8,
            activation="relu"
        ),

        Dense(
            1,
            activation="linear"
        )

    ])

    ann_model.compile(
        optimizer=Adam(
            learning_rate=0.001
        ),
        loss="mse",
        metrics=["mae"]
    )

    history = ann_model.fit(
        X_scaled,
        y_scaled,
        epochs=300,
        batch_size=1,
        verbose=0
    )

    # Save models
    joblib.dump(
        x_scaler,
        x_scaler_path
    )

    joblib.dump(
        y_scaler,
        y_scaler_path
    )

    joblib.dump(
        lr_model,
        lr_model_path
    )

    ann_model.save(
        ann_model_path
    )

    print("\nModels trained and saved.")

else:

    print("\nExisting models loaded successfully.")

    # Recompile the ANN for evaluation.
    ann_model.compile(
        optimizer=Adam(
            learning_rate=0.001
        ),
        loss="mse",
        metrics=["mae"]
    )

    history = None


# ============================================================
# 7. GENERATE MODEL PREDICTIONS
# ============================================================

X_scaled = x_scaler.transform(X)

lr_predictions = lr_model.predict(
    X_scaled
)

ann_predictions_scaled = ann_model.predict(
    X_scaled,
    verbose=0
)

ann_predictions = y_scaler.inverse_transform(
    ann_predictions_scaled
).flatten()

print("\nLinear Regression Predictions:")
print(lr_predictions)

print("\nANN Predictions:")
print(ann_predictions)


# ============================================================
# 8. MODEL EVALUATION
# ============================================================

def calculate_metrics(
    actual,
    predicted
):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual,
        predicted
    )

    # MAPE is undefined when actual values are zero.
    nonzero = actual != 0

    if np.any(nonzero):

        mape = np.mean(
            np.abs(
                (
                    actual[nonzero]
                    - predicted[nonzero]
                )
                / actual[nonzero]
            )
        ) * 100

        accuracy = 100 - mape

    else:

        mape = np.nan
        accuracy = np.nan

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": float(mape),
        "Accuracy_Percentage": float(accuracy)
    }


lr_metrics = calculate_metrics(
    y,
    lr_predictions
)

ann_metrics = calculate_metrics(
    y,
    ann_predictions
)

print("\nLinear Regression Metrics:")
print(lr_metrics)

print("\nANN Metrics:")
print(ann_metrics)

print(
    "\nIMPORTANT: These are training-set fit metrics. "
    "They are not validated forecasting accuracy."
)


# ============================================================
# 9. SAVE MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame({

    "Model": [
        "Linear Regression",
        "ANN"
    ],

    "MAE": [
        lr_metrics["MAE"],
        ann_metrics["MAE"]
    ],

    "MSE": [
        lr_metrics["MSE"],
        ann_metrics["MSE"]
    ],

    "RMSE": [
        lr_metrics["RMSE"],
        ann_metrics["RMSE"]
    ],

    "R2": [
        lr_metrics["R2"],
        ann_metrics["R2"]
    ],

    "MAPE": [
        lr_metrics["MAPE"],
        ann_metrics["MAPE"]
    ],

    "Accuracy_Percentage": [
        lr_metrics["Accuracy_Percentage"],
        ann_metrics["Accuracy_Percentage"]
    ]
})

comparison_df.to_csv(
    os.path.join(
        PROJECT_PATH,
        "model_comparison.csv"
    ),
    index=False
)

print("\nModel Comparison:")
print(comparison_df)


# ============================================================
# 10. ACCURACY GRAPH
# ============================================================

# Here "accuracy" is defined as 100 - training MAPE.
# This is not classification accuracy or test accuracy.

plt.figure(figsize=(10, 6))

models = [
    "Linear Regression",
    "ANN"
]

accuracies = [
    lr_metrics["Accuracy_Percentage"],
    ann_metrics["Accuracy_Percentage"]
]

bars = plt.bar(
    models,
    accuracies,
    color=["royalblue", "darkorange"],
    width=0.5
)

for bar, accuracy in zip(
    bars,
    accuracies
):

    if np.isfinite(accuracy):

        plt.text(
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height(),
            f"{accuracy:.2f}%",
            ha="center",
            va="bottom",
            fontsize=12
        )

plt.title(
    "Model Training Fit Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Machine Learning Models")
plt.ylabel("100 - Training MAPE (%)")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_PATH,
        "accuracy_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nAccuracy Graph Saved.")


# ============================================================
# 11. CORRELATION HEATMAP
# ============================================================

heatmap_data = data.copy()

heatmap_data["Budget_Growth"] = (
    heatmap_data[
        "Budget_Allocation"
    ].pct_change() * 100
)

correlation_matrix = heatmap_data[
    [
        "Year",
        "Budget_Allocation",
        "Budget_Growth"
    ]
].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True,
    vmin=-1,
    vmax=1
)

plt.title(
    "Government Budget Allocation Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_PATH,
        "heatmap.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nHeatmap Saved.")


# ============================================================
# 12. MODEL COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

x_positions = np.arange(
    len(models)
)

width = 0.35

plt.bar(
    x_positions - width / 2,
    [
        lr_metrics["MAE"],
        ann_metrics["MAE"]
    ],
    width,
    label="MAE",
    color="steelblue"
)

plt.bar(
    x_positions + width / 2,
    [
        lr_metrics["RMSE"],
        ann_metrics["RMSE"]
    ],
    width,
    label="RMSE",
    color="coral"
)

plt.xticks(
    x_positions,
    models
)

plt.title(
    "Machine Learning Model Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")
plt.ylabel("Training Error")

plt.legend()
plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_PATH,
        "comparison_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nComparison Graph Saved.")


# ============================================================
# 13. GENERATE RESULT CSV
# ============================================================

result_df = pd.DataFrame({

    "Year": data["Year"],

    "Actual_Budget": y,

    "Linear_Regression_Prediction":
        lr_predictions,

    "ANN_Prediction":
        ann_predictions

})

result_df["LR_Absolute_Error"] = np.abs(
    result_df["Actual_Budget"]
    - result_df["Linear_Regression_Prediction"]
)

result_df["ANN_Absolute_Error"] = np.abs(
    result_df["Actual_Budget"]
    - result_df["ANN_Prediction"]
)

result_path = os.path.join(
    PROJECT_PATH,
    "result.csv"
)

result_df.to_csv(
    result_path,
    index=False
)

print("\nResult CSV:")
print(result_df)

print("\nSaved:", result_path)


# ============================================================
# 14. RESULT GRAPH
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(
    result_df["Year"],
    result_df["Actual_Budget"],
    marker="o",
    linewidth=3,
    color="black",
    label="Actual Budget"
)

plt.plot(
    result_df["Year"],
    result_df[
        "Linear_Regression_Prediction"
    ],
    marker="s",
    linestyle="--",
    linewidth=2,
    color="royalblue",
    label="Linear Regression"
)

plt.plot(
    result_df["Year"],
    result_df["ANN_Prediction"],
    marker="^",
    linestyle="--",
    linewidth=2,
    color="darkorange",
    label="ANN"
)

plt.title(
    "Actual vs Predicted Government Budget Allocation",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Budget Allocation")

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_PATH,
        "result_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nResult Graph Saved.")


# ============================================================
# 15. FUTURE BUDGET PREDICTION
# ============================================================

last_year = int(
    data["Year"].max()
)

future_years = np.arange(
    last_year + 1,
    last_year + 7
).reshape(-1, 1)

future_scaled = x_scaler.transform(
    future_years
)

# Linear Regression Forecast
future_lr = lr_model.predict(
    future_scaled
)

# ANN Forecast
future_ann_scaled = ann_model.predict(
    future_scaled,
    verbose=0
)

future_ann = y_scaler.inverse_transform(
    future_ann_scaled
).flatten()

prediction_df = pd.DataFrame({

    "Year": future_years.flatten(),

    "Linear_Regression_Prediction":
        future_lr,

    "ANN_Prediction":
        future_ann

})

prediction_df["Financial_Year"] = (
    prediction_df["Year"].astype(str)
    + "-"
    + (
        (
            prediction_df["Year"] + 1
        ) % 100
    ).astype(str).str.zfill(2)
)

prediction_df = prediction_df[
    [
        "Financial_Year",
        "Year",
        "Linear_Regression_Prediction",
        "ANN_Prediction"
    ]
]

prediction_path = os.path.join(
    PROJECT_PATH,
    "prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

print("\nFuture Budget Predictions:")
print(prediction_df)

print("\nSaved:", prediction_path)


# ============================================================
# 16. PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(12, 7))

plt.plot(
    data["Year"],
    data["Budget_Allocation"],
    marker="o",
    linewidth=3,
    color="black",
    label="Historical Budget"
)

plt.plot(
    prediction_df["Year"],
    prediction_df[
        "Linear_Regression_Prediction"
    ],
    marker="s",
    linestyle="--",
    linewidth=2,
    color="royalblue",
    label="Linear Regression Forecast"
)

plt.plot(
    prediction_df["Year"],
    prediction_df["ANN_Prediction"],
    marker="^",
    linestyle="--",
    linewidth=2,
    color="darkorange",
    label="ANN Forecast"
)

plt.axvline(
    x=last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)

plt.title(
    "Future Government Budget Allocation Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")
plt.ylabel("Budget Allocation")

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        PROJECT_PATH,
        "prediction_graph.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nPrediction Graph Saved.")


# ============================================================
# 17. ANN TRAINING HISTORY
# ============================================================

# Training history is available only if the ANN
# was trained during this execution.

if history is not None:

    training_df = pd.DataFrame({
        "Epoch": np.arange(
            1,
            len(history.history["loss"]) + 1
        ),
        "Loss": history.history["loss"],
        "MAE": history.history["mae"]
    })

    training_df.to_csv(
        os.path.join(
            PROJECT_PATH,
            "training_history.csv"
        ),
        index=False
    )

    plt.figure(figsize=(10, 6))

    plt.plot(
        training_df["Epoch"],
        training_df["Loss"],
        label="Training Loss",
        color="purple",
        linewidth=2
    )

    plt.title(
        "ANN Training Loss",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel("Epoch")
    plt.ylabel("Mean Squared Error")

    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            PROJECT_PATH,
            "ann_loss_graph.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    print("\nANN Training History Saved.")

else:

    print(
        "\nANN training history unavailable because "
        "an existing trained model was loaded."
    )


# ============================================================
# 18. SAVE EVALUATION METRICS AS JSON
# ============================================================

metrics_data = {

    "project":
        "AI-Based Government Budget Allocation "
        "Trend Analysis and Forecasting",

    "evaluation_type":
        "In-sample training evaluation",

    "linear_regression":
        lr_metrics,

    "artificial_neural_network":
        ann_metrics,

    "dataset_records":
        int(len(data)),

    "forecast_years":
        int(len(prediction_df)),

    "warning":
        "The dataset is too small for reliable "
        "forecast validation. Predictions are illustrative."
}

with open(
    os.path.join(
        PROJECT_PATH,
        "evaluation_metrics.json"
    ),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metrics_data,
        file,
        indent=4
    )

print("\nEvaluation Metrics JSON Saved.")


# ============================================================
# 19. FINAL OUTPUT SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("ALL RESULTS GENERATED SUCCESSFULLY")
print("=" * 65)

generated_files = [

    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "result.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "model_comparison.csv",
    "evaluation_metrics.json"
]

if history is not None:

    generated_files.extend([
        "training_history.csv",
        "ann_loss_graph.png"
    ])

print("\nGenerated Files:\n")

for filename in generated_files:

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    if os.path.exists(filepath):

        print("SUCCESS:", filename)

    else:

        print("MISSING:", filename)

print("\nProject Directory:")
print(PROJECT_PATH)

print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED")
print("=" * 65)
```
