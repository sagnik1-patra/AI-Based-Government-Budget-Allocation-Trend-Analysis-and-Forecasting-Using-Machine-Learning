```python
import os
import re
import json
import yaml
import joblib
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

np.random.seed(42)

sns.set_theme(style="whitegrid")


# ============================================================
# 2. PROJECT DIRECTORY
# ============================================================

PROJECT_PATH = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Government Budget Allocation Trend Analysis "
    r"and Forecasting Using Machine Learning"
)

DATASET_PATH = os.path.join(
    PROJECT_PATH,
    "RS_Session_252_AU_600.csv"
)

os.makedirs(
    PROJECT_PATH,
    exist_ok=True
)

PREFIX = "pso_"


def save_path(filename):

    return os.path.join(
        PROJECT_PATH,
        PREFIX + filename
    )


print("=" * 65)
print("PARTICLE SWARM OPTIMIZATION - BUDGET FORECASTING")
print("=" * 65)

print("\nProject Directory:")
print(PROJECT_PATH)


# ============================================================
# 3. LOAD DATASET
# ============================================================

if not os.path.exists(DATASET_PATH):

    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

df = pd.read_csv(DATASET_PATH)

df.columns = (
    df.columns.astype(str).str.strip()
)

print("\nOriginal Dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Columns:")
print(df.columns.tolist())


# ============================================================
# 4. DATA PREPROCESSING
# ============================================================

# Supports:
#
# 1. Long format:
#    Financial Year | Budget Allocation
#
# 2. Wide format:
#    2017-18 | 2018-19 | 2019-20
#
# If multiple budget categories exist, select the
# required category before running the model.


year_pattern = re.compile(
    r"((?:19|20)\d{2})\s*[-–/]\s*"
    r"(?:\d{2}|(?:19|20)\d{2})"
)


def extract_year(value):

    text = str(value).strip()

    match = year_pattern.search(text)

    if match:
        return int(match.group(1))

    if re.fullmatch(
        r"(?:19|20)\d{2}",
        text
    ):
        return int(text)

    return np.nan


def convert_numeric(series):

    cleaned = (
        series.astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("₹", "", regex=False)
        .str.replace(
            r"(?i)\s*crores?\s*",
            "",
            regex=True
        )
        .str.strip()
    )

    return pd.to_numeric(
        cleaned,
        errors="coerce"
    )


year_columns = [
    col for col in df.columns
    if year_pattern.search(str(col))
]


if len(year_columns) >= 2:

    numeric_data = df[
        year_columns
    ].apply(convert_numeric)

    best_row = (
        numeric_data.notna()
        .sum(axis=1)
        .idxmax()
    )

    records = []

    for col in year_columns:

        value = numeric_data.loc[
            best_row,
            col
        ]

        if pd.notna(value):

            records.append({

                "Year":
                    extract_year(col),

                "Budget_Allocation":
                    float(value)
            })

    data = pd.DataFrame(records)


else:

    year_candidates = [
        col for col in df.columns
        if "year" in col.lower()
    ]

    budget_candidates = [
        col for col in df.columns
        if any(
            word in col.lower()
            for word in [
                "allocation",
                "budget",
                "amount",
                "fund"
            ]
        )
    ]

    if not year_candidates:

        raise ValueError(
            "Financial year column not found."
        )

    if not budget_candidates:

        raise ValueError(
            "Budget allocation column not found."
        )

    year_col = year_candidates[0]

    budget_col = next(
        (
            col for col in budget_candidates
            if col != year_col
        ),
        None
    )

    if budget_col is None:

        raise ValueError(
            "Separate budget allocation column not found."
        )

    data = pd.DataFrame({

        "Year":
            df[year_col].apply(
                extract_year
            ),

        "Budget_Allocation":
            convert_numeric(
                df[budget_col]
            )
    })


data = data.dropna()

data = data.sort_values(
    "Year"
).reset_index(drop=True)

if data["Year"].duplicated().any():

    raise ValueError(
        "Multiple budget values found for the same year. "
        "Filter the dataset to one budget category."
    )

if len(data) < 2:

    raise ValueError(
        "At least two historical records are required."
    )


data.to_csv(
    save_path("cleaned_dataset.csv"),
    index=False
)

print("\nCleaned Dataset:")
print(data)


# ============================================================
# 5. FEATURE PREPARATION
# ============================================================

years = data[
    "Year"
].to_numpy(dtype=float)

y = data[
    "Budget_Allocation"
].to_numpy(dtype=float)


# Normalize years
year_center = float(
    np.mean(years)
)

year_scale = float(
    np.std(years)
)

if year_scale == 0:
    year_scale = 1.0


X_scaled = (
    years - year_center
) / year_scale


# Normalize target
target_center = float(
    np.mean(y)
)

target_scale = float(
    np.std(y)
)

if target_scale == 0:
    target_scale = 1.0


y_scaled = (
    y - target_center
) / target_scale


print("\nNormalized Features:")
print(X_scaled)

print("\nNormalized Target:")
print(y_scaled)


# ============================================================
# 6. PSO MODEL REPRESENTATION
# ============================================================

# Each particle represents two regression coefficients:
#
# particle = [a, b]
#
# Prediction equation:
#
# y_scaled = a + b * x_scaled
#
# PSO minimizes the Mean Squared Error.


def predict_scaled(
    particles,
    x_values
):

    particles = np.asarray(
        particles,
        dtype=float
    )

    x_values = np.asarray(
        x_values,
        dtype=float
    )

    return (
        particles[..., 0, None]
        +
        particles[..., 1, None]
        * x_values
    )


def fitness_function(
    particles,
    x_values,
    targets
):

    predictions = predict_scaled(
        particles,
        x_values
    )

    errors = (
        predictions - targets
    ) ** 2

    return np.mean(
        errors,
        axis=-1
    )


# ============================================================
# 7. PARTICLE SWARM OPTIMIZATION ALGORITHM
# ============================================================

def particle_swarm_optimization(

    x_values,
    targets,

    num_particles=50,
    iterations=200,

    inertia_weight=0.7,
    cognitive_coefficient=1.5,
    social_coefficient=1.5,

    lower_bound=-5,
    upper_bound=5,

    random_state=42
):

    rng = np.random.default_rng(
        random_state
    )

    dimensions = 2

    # Initialize particle positions
    positions = rng.uniform(
        lower_bound,
        upper_bound,
        size=(
            num_particles,
            dimensions
        )
    )

    # Initialize velocities
    velocities = rng.uniform(
        -0.5,
        0.5,
        size=(
            num_particles,
            dimensions
        )
    )

    # Evaluate initial fitness
    fitness = fitness_function(
        positions,
        x_values,
        targets
    )

    # Personal best positions
    personal_best_positions = (
        positions.copy()
    )

    personal_best_fitness = (
        fitness.copy()
    )

    # Global best
    best_index = np.argmin(
        fitness
    )

    global_best_position = (
        positions[best_index].copy()
    )

    global_best_fitness = float(
        fitness[best_index]
    )

    convergence_history = []

    # PSO iterations
    for iteration in range(
        iterations
    ):

        r1 = rng.random(
            size=positions.shape
        )

        r2 = rng.random(
            size=positions.shape
        )

        # Update particle velocities
        velocities = (

            inertia_weight
            * velocities

            +

            cognitive_coefficient
            * r1
            * (
                personal_best_positions
                - positions
            )

            +

            social_coefficient
            * r2
            * (
                global_best_position
                - positions
            )
        )

        # Update particle positions
        positions = (
            positions + velocities
        )

        # Apply position bounds
        positions = np.clip(
            positions,
            lower_bound,
            upper_bound
        )

        # Evaluate updated positions
        fitness = fitness_function(
            positions,
            x_values,
            targets
        )

        # Update personal bests
        improved = (
            fitness
            < personal_best_fitness
        )

        personal_best_positions[
            improved
        ] = positions[
            improved
        ]

        personal_best_fitness[
            improved
        ] = fitness[
            improved
        ]

        # Update global best
        current_best_index = np.argmin(
            personal_best_fitness
        )

        current_best_fitness = (
            personal_best_fitness[
                current_best_index
            ]
        )

        if (
            current_best_fitness
            < global_best_fitness
        ):

            global_best_fitness = float(
                current_best_fitness
            )

            global_best_position = (
                personal_best_positions[
                    current_best_index
                ].copy()
            )

        convergence_history.append(
            global_best_fitness
        )

        if (
            iteration + 1
        ) % 20 == 0:

            print(
                f"Iteration {iteration + 1:3d} "
                f"| Best MSE: "
                f"{global_best_fitness:.8f}"
            )

    return (
        global_best_position,
        global_best_fitness,
        convergence_history
    )


# ============================================================
# 8. TRAIN PSO MODEL
# ============================================================

PSO_PARTICLES = 50

PSO_ITERATIONS = 200

PSO_INERTIA = 0.7

PSO_COGNITIVE = 1.5

PSO_SOCIAL = 1.5


print("\nTraining PSO Model...\n")

(
    best_particle,
    best_fitness,
    convergence_history
) = particle_swarm_optimization(

    X_scaled,
    y_scaled,

    num_particles=PSO_PARTICLES,

    iterations=PSO_ITERATIONS,

    inertia_weight=PSO_INERTIA,

    cognitive_coefficient=PSO_COGNITIVE,

    social_coefficient=PSO_SOCIAL
)


print("\nPSO Training Completed.")

print("\nBest Particle:")
print(best_particle)

print("\nBest Fitness:")
print(best_fitness)


# ============================================================
# 9. GENERATE PSO PREDICTIONS
# ============================================================

pso_scaled_predictions = predict_scaled(
    best_particle,
    X_scaled
)

pso_predictions = (
    pso_scaled_predictions
    * target_scale
    + target_center
)

print("\nPSO Predictions:")
print(pso_predictions)


# ============================================================
# 10. BASELINE LINEAR REGRESSION
# ============================================================

baseline_model = LinearRegression()

baseline_model.fit(
    X_scaled.reshape(-1, 1),
    y
)

baseline_predictions = baseline_model.predict(
    X_scaled.reshape(-1, 1)
)

print("\nBaseline Predictions:")
print(baseline_predictions)


# ============================================================
# 11. MODEL EVALUATION
# ============================================================

def evaluate_model(
    actual,
    predicted
):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        actual,
        predicted
    )

    nonzero = (
        actual != 0
    )

    if np.any(nonzero):

        mape = np.mean(
            np.abs(
                (
                    actual[nonzero]
                    - predicted[nonzero]
                )
                / actual[nonzero]
            )
        ) * 100

        accuracy = 100 - mape

    else:

        mape = np.nan
        accuracy = np.nan

    return {

        "MAE":
            float(mae),

        "MSE":
            float(mse),

        "RMSE":
            float(rmse),

        "R2":
            float(r2),

        "MAPE":
            float(mape),

        "Accuracy_Percentage":
            float(accuracy)
    }


pso_metrics = evaluate_model(
    y,
    pso_predictions
)

baseline_metrics = evaluate_model(
    y,
    baseline_predictions
)


print("\nPSO Training Metrics:")
print(pso_metrics)

print("\nBaseline Training Metrics:")
print(baseline_metrics)

print(
    "\nWARNING: Metrics are calculated on training "
    "data and do not measure future forecasting accuracy."
)


# ============================================================
# 12. SAVE PSO MODEL
# ============================================================

pso_model_data = {

    "algorithm":
        "Particle Swarm Optimization",

    "model_type":
        "Linear Regression",

    "best_particle":
        best_particle,

    "year_center":
        year_center,

    "year_scale":
        year_scale,

    "target_center":
        target_center,

    "target_scale":
        target_scale,

    "best_fitness":
        best_fitness
}


joblib.dump(
    pso_model_data,
    save_path("model.pkl")
)

print("\nPSO Model Saved.")


# ============================================================
# 13. SAVE PSO CONFIGURATION YAML
# ============================================================

pso_config = {

    "project":
        "Government Budget Allocation Forecasting",

    "algorithm":
        "Particle Swarm Optimization",

    "model":
        "Linear Trend",

    "particles":
        PSO_PARTICLES,

    "iterations":
        PSO_ITERATIONS,

    "inertia_weight":
        PSO_INERTIA,

    "cognitive_coefficient":
        PSO_COGNITIVE,

    "social_coefficient":
        PSO_SOCIAL,

    "features": [
        "Year"
    ],

    "target":
        "Budget_Allocation",

    "evaluation":
        "Training Data Only"
}


with open(
    save_path("model_config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        pso_config,
        file,
        sort_keys=False
    )

print("\nPSO YAML Saved.")


# ============================================================
# 14. SAVE PSO METADATA JSON
# ============================================================

pso_metadata = {

    "algorithm":
        "Particle Swarm Optimization",

    "dataset":
        "RS_Session_252_AU_600.csv",

    "total_records":
        int(len(data)),

    "best_particle":
        best_particle.tolist(),

    "best_fitness":
        float(best_fitness),

    "metrics":
        pso_metrics,

    "limitations":
        "The dataset is too small for reliable "
        "future forecasting."
}


with open(
    save_path("model_metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        pso_metadata,
        file,
        indent=4
    )

print("\nPSO Metadata Saved.")


# ============================================================
# 15. PSO ACCURACY GRAPH
# ============================================================

# This graph displays 100 - training MAPE.
# It is not classification accuracy or test accuracy.

models = [
    "Linear Regression",
    "PSO Optimized Regression"
]

accuracies = [
    baseline_metrics["Accuracy_Percentage"],
    pso_metrics["Accuracy_Percentage"]
]


plt.figure(figsize=(10, 6))

bars = plt.bar(
    models,
    accuracies,
    color=[
        "steelblue",
        "darkorange"
    ],
    width=0.5
)


for bar, value in zip(
    bars,
    accuracies
):

    if np.isfinite(value):

        plt.text(
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height(),
            f"{value:.2f}%",
            ha="center",
            va="bottom",
            fontsize=12
        )


plt.title(
    "PSO Training Fit Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")

plt.ylabel(
    "100 - Training MAPE (%)"
)

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    save_path("accuracy_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nPSO Accuracy Graph Saved.")


# ============================================================
# 16. PSO CORRELATION HEATMAP
# ============================================================

heatmap_df = data.copy()

heatmap_df["PSO_Prediction"] = (
    pso_predictions
)

heatmap_df["Budget_Growth"] = (
    heatmap_df[
        "Budget_Allocation"
    ].pct_change() * 100
)


correlation_matrix = heatmap_df[
    [
        "Year",
        "Budget_Allocation",
        "PSO_Prediction",
        "Budget_Growth"
    ]
].corr()


plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title(
    "PSO Budget Allocation Correlation Heatmap",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    save_path("heatmap.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nPSO Heatmap Saved.")


# ============================================================
# 17. PSO MODEL COMPARISON CSV
# ============================================================

comparison_df = pd.DataFrame({

    "Model":
        models,

    "MAE": [
        baseline_metrics["MAE"],
        pso_metrics["MAE"]
    ],

    "MSE": [
        baseline_metrics["MSE"],
        pso_metrics["MSE"]
    ],

    "RMSE": [
        baseline_metrics["RMSE"],
        pso_metrics["RMSE"]
    ],

    "R2": [
        baseline_metrics["R2"],
        pso_metrics["R2"]
    ],

    "MAPE": [
        baseline_metrics["MAPE"],
        pso_metrics["MAPE"]
    ],

    "Accuracy_Percentage": [
        baseline_metrics["Accuracy_Percentage"],
        pso_metrics["Accuracy_Percentage"]
    ]
})


comparison_df.to_csv(
    save_path("model_comparison.csv"),
    index=False
)

print("\nPSO Model Comparison:")
print(comparison_df)


# ============================================================
# 18. PSO COMPARISON GRAPH
# ============================================================

x_positions = np.arange(
    len(models)
)

width = 0.35


plt.figure(figsize=(11, 6))


plt.bar(
    x_positions - width / 2,
    comparison_df["MAE"],
    width,
    label="MAE",
    color="royalblue"
)


plt.bar(
    x_positions + width / 2,
    comparison_df["RMSE"],
    width,
    label="RMSE",
    color="darkorange"
)


plt.xticks(
    x_positions,
    models
)


plt.title(
    "PSO vs Linear Regression - Training Error",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Models")

plt.ylabel("Error")

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    save_path("comparison_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nPSO Comparison Graph Saved.")


# ============================================================
# 19. PSO RESULT CSV
# ============================================================

result_df = pd.DataFrame({

    "Year":
        years.astype(int),

    "Actual_Budget":
        y,

    "PSO_Prediction":
        pso_predictions,

    "Baseline_Prediction":
        baseline_predictions
})


result_df["PSO_Absolute_Error"] = np.abs(
    result_df["Actual_Budget"]
    - result_df["PSO_Prediction"]
)


result_df["Baseline_Absolute_Error"] = np.abs(
    result_df["Actual_Budget"]
    - result_df["Baseline_Prediction"]
)


result_df.to_csv(
    save_path("result.csv"),
    index=False
)

print("\nPSO Result CSV:")
print(result_df)


# ============================================================
# 20. PSO RESULT GRAPH
# ============================================================

plt.figure(figsize=(12, 7))


plt.plot(
    result_df["Year"],
    result_df["Actual_Budget"],
    marker="o",
    linewidth=3,
    color="black",
    label="Actual Budget"
)


plt.plot(
    result_df["Year"],
    result_df["PSO_Prediction"],
    marker="s",
    linestyle="--",
    linewidth=2,
    color="darkorange",
    label="PSO Prediction"
)


plt.plot(
    result_df["Year"],
    result_df["Baseline_Prediction"],
    marker="^",
    linestyle=":",
    linewidth=2,
    color="royalblue",
    label="Baseline Prediction"
)


plt.title(
    "PSO Actual vs Predicted Budget Allocation",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")

plt.ylabel("Budget Allocation")

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    save_path("result_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nPSO Result Graph Saved.")


# ============================================================
# 21. PSO FUTURE BUDGET PREDICTION
# ============================================================

last_year = int(
    np.max(years)
)


future_years = np.arange(
    last_year + 1,
    last_year + 7
)


future_scaled = (
    future_years - year_center
) / year_scale


future_pso_scaled = predict_scaled(
    best_particle,
    future_scaled
)


future_pso = (
    future_pso_scaled
    * target_scale
    + target_center
)


future_baseline = baseline_model.predict(
    future_scaled.reshape(-1, 1)
)


prediction_df = pd.DataFrame({

    "Year":
        future_years,

    "PSO_Prediction":
        future_pso,

    "Baseline_Prediction":
        future_baseline
})


prediction_df["Financial_Year"] = (
    prediction_df["Year"].astype(str)
    + "-"
    + (
        (
            prediction_df["Year"] + 1
        ) % 100
    ).astype(str).str.zfill(2)
)


prediction_df = prediction_df[
    [
        "Financial_Year",
        "Year",
        "PSO_Prediction",
        "Baseline_Prediction"
    ]
]


prediction_df.to_csv(
    save_path("prediction.csv"),
    index=False
)


print("\nPSO Future Predictions:")
print(prediction_df)


# ============================================================
# 22. PSO PREDICTION GRAPH
# ============================================================

plt.figure(figsize=(12, 7))


plt.plot(
    years,
    y,
    marker="o",
    linewidth=3,
    color="black",
    label="Historical Budget"
)


plt.plot(
    future_years,
    future_pso,
    marker="s",
    linestyle="--",
    linewidth=2,
    color="darkorange",
    label="PSO Forecast"
)


plt.plot(
    future_years,
    future_baseline,
    marker="^",
    linestyle=":",
    linewidth=2,
    color="royalblue",
    label="Linear Regression Forecast"
)


plt.axvline(
    x=last_year,
    color="gray",
    linestyle=":",
    label="Forecast Boundary"
)


plt.title(
    "PSO Government Budget Allocation Forecast",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Financial Year Starting")

plt.ylabel("Predicted Budget Allocation")

plt.legend()

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    save_path("prediction_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nPSO Prediction Graph Saved.")


# ============================================================
# 23. PSO CONVERGENCE CSV
# ============================================================

convergence_df = pd.DataFrame({

    "Iteration": np.arange(
        1,
        len(convergence_history) + 1
    ),

    "Best_Fitness":
        convergence_history
})


convergence_df.to_csv(
    save_path("convergence.csv"),
    index=False
)


print("\nPSO Convergence Data:")
print(convergence_df.tail())


# ============================================================
# 24. PSO CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(11, 6))


plt.plot(
    convergence_df["Iteration"],
    convergence_df["Best_Fitness"],
    color="purple",
    linewidth=2
)


plt.title(
    "PSO Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Iteration")

plt.ylabel("Best Fitness (MSE)")

plt.grid(
    True,
    alpha=0.3
)

plt.tight_layout()

plt.savefig(
    save_path("convergence_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nPSO Convergence Graph Saved.")


# ============================================================
# 25. SAVE PSO EVALUATION METRICS
# ============================================================

evaluation_data = {

    "algorithm":
        "Particle Swarm Optimization",

    "dataset_records":
        int(len(data)),

    "PSO_Metrics":
        pso_metrics,

    "Baseline_Metrics":
        baseline_metrics,

    "Best_Fitness":
        float(best_fitness),

    "Evaluation_Type":
        "In-sample training",

    "Warning":
        "The dataset is too small for reliable "
        "forecast validation."
}


with open(
    save_path("evaluation_metrics.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        evaluation_data,
        file,
        indent=4
    )

print("\nPSO Evaluation Metrics Saved.")


# ============================================================
# 26. FINAL OUTPUT SUMMARY
# ============================================================

generated_files = [

    "pso_model.pkl",

    "pso_model_config.yaml",

    "pso_model_metadata.json",

    "pso_cleaned_dataset.csv",

    "pso_accuracy_graph.png",

    "pso_heatmap.png",

    "pso_comparison_graph.png",

    "pso_model_comparison.csv",

    "pso_result.csv",

    "pso_result_graph.png",

    "pso_prediction.csv",

    "pso_prediction_graph.png",

    "pso_convergence.csv",

    "pso_convergence_graph.png",

    "pso_evaluation_metrics.json"
]


print("\n" + "=" * 65)

print("PSO PROJECT EXECUTION COMPLETED")

print("=" * 65)


for filename in generated_files:

    filepath = os.path.join(
        PROJECT_PATH,
        filename
    )

    if os.path.exists(filepath):

        print("SUCCESS:", filename)

    else:

        print("MISSING:", filename)


print("\nAll PSO results saved in:")

print(PROJECT_PATH)

print("\n" + "=" * 65)
```
